```python
# 2️⃣ Bottom 10 States
plt.figure(figsize=(8,6))
df['state'].value_counts().tail(10).plot(kind='bar')
plt.title("Bottom 10 States by Number of Airports")
plt.xlabel("State")
plt.ylabel("Number of Airports")
plt.xticks(rotation=45)
plt.show()
```

```python
# 3️⃣ Pie Chart – Top 5 States Share
top5 = df['state'].value_counts().head(5)

plt.figure(figsize=(6,6))
plt.pie(top5, labels=top5.index, autopct='%1.1f%%')
plt.title("Top 5 States Airport Share")
plt.show()
```

```python
# 4️⃣ Histogram – Airport Name Length
plt.figure(figsize=(8,6))
plt.hist(df['name_length'], bins=20)
plt.title("Distribution of Airport Name Length")
plt.xlabel("Name Length")
plt.ylabel("Frequency")
plt.show()
```

```python
# 5️⃣ Boxplot – Name Length (Outlier Detection)
plt.figure(figsize=(6,6))
plt.boxplot(df['name_length'])
plt.title("Boxplot of Airport Name Length")
plt.ylabel("Name Length")
plt.show()
```

```python
# 6️⃣ Airports Per City Distribution
city_counts = df['city'].value_counts()

plt.figure(figsize=(8,6))
plt.hist(city_counts, bins=10)
plt.title("Distribution of Airports per City")
plt.xlabel("Number of Airports in a City")
plt.ylabel("Frequency")
plt.show()
```

```python
# 7️⃣ Cumulative Distribution of Airport IDs
sorted_ids = df['airport_id'].sort_values()

plt.figure(figsize=(8,6))
plt.plot(sorted_ids.values)
plt.title("Cumulative Distribution of Airport IDs")
plt.xlabel("Index")
plt.ylabel("Airport ID")
plt.show()
```
